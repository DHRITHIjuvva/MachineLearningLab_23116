```python
# Lab04_all_questions.py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report, mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import make_blobs
from typing import Tuple, Dict, Any
import math

def load_data(path="/mnt/data/extracted_audio_features.csv"):
    df = pd.read_csv('/Users/dhrithijuvva/Desktop/Women_ML/extracted_audio_features.csv')
    feature_cols = [c for c in df.columns if c.startswith("MFCC")]
    X = df[feature_cols].values
    y = df["Label"].values
    return df, X, y, feature_cols

# ---------- A1: confusion matrix & metrics ----------
def compute_confusion_metrics(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    report = classification_report(y_true, y_pred, zero_division=0, output_dict=True)
    return {"confusion_matrix": cm, "report": report}

# ---------- A2: MSE RMSE MAPE R2 ----------
def regression_metrics(y_true, y_pred):
    """
    Computes regression metrics. If shapes mismatch, raise ValueError.
    """
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    if y_true.shape != y_pred.shape:
        raise ValueError("Shapes of true and pred must match")
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    # MAPE: handle zeros in y_true carefully
    denom = np.where(y_true == 0, np.nan, y_true)
    mape = np.nanmean(np.abs((y_true - y_pred) / denom)) * 100.0
    r2 = r2_score(y_true, y_pred)
    return {"MSE": mse, "RMSE": rmse, "MAPE_percent": mape, "R2": r2}

# ---------- A3-A6: generate synthetic train data and visualize kNN boundaries ----------
def generate_2d_train(n_samples=20, low=1, high=10, random_state=42):
    rng = np.random.RandomState(random_state)
    X = rng.uniform(low, high, size=(n_samples,2))
    # create labels based on a simple rule or randomly assign half to 0 half to1
    y = np.zeros(n_samples, dtype=int)
    y[:n_samples//2] = 0
    y[n_samples//2:] = 1
    # shuffle
    perm = rng.permutation(n_samples)
    return X[perm], y[perm]

def classify_grid_with_knn(train_X, train_y, k=3, grid_step=0.1, bounds=(0,10,0,10)):
    x_min, x_max, y_min, y_max = bounds[0], bounds[1], bounds[2], bounds[3] if isinstance(bounds, tuple) and len(bounds)==4 else (0,10,0,10)
    xx = np.arange(x_min, x_max, grid_step)
    yy = np.arange(y_min, y_max, grid_step)
    xxg, yyg = np.meshgrid(xx, yy)
    grid = np.c_[xxg.ravel(), yyg.ravel()]
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(train_X, train_y)
    preds = knn.predict(grid)
    Z = preds.reshape(xxg.shape)
    return xxg, yyg, Z

# ---------- A7: GridSearch/RandomSearch to find best k ----------
def find_best_k_via_grid(X_train, y_train, k_values=[1,3,5,7,9]):
    param_grid = {"n_neighbors": k_values}
    knn = KNeighborsClassifier()
    gs = GridSearchCV(knn, param_grid, cv=5, scoring='accuracy')
    gs.fit(X_train, y_train)
    return gs

# ---------- Main ----------
if __name__ == "__main__":
    df, X, y, feature_cols = load_data()
    # A1: compute confusion for an example: use two-class split and k=3
    # pick two classes automatically
    labels = np.unique(y)
    if len(labels) >= 2:
        classes = (labels[0], labels[1])
        mask = np.isin(y, classes)
        X2 = X[mask]
        y2 = y[mask]
        mapping = {classes[0]:0, classes[1]:1}
        y2_enc = np.vectorize(mapping.get)(y2)
        Xtr, Xte, ytr, yte = train_test_split(X2, y2_enc, test_size=0.3, random_state=42, stratify=y2_enc)
        knn = KNeighborsClassifier(n_neighbors=3)
        knn.fit(Xtr, ytr)
        ypred = knn.predict(Xte)
        cm_metrics = compute_confusion_metrics(yte, ypred)
        print("Lab04 A1 confusion matrix:\n", cm_metrics["confusion_matrix"])
        print("Lab04 A1 classification report (selected metrics):")
        # pretty print a few fields
        for k, v in cm_metrics["report"].items():
            if k in ['0','1','macro avg','weighted avg']:
                print(k, v)
    else:
        print("Not enough classes for A1.")

    # A2: regression metrics - fallback: no Lab02 data present.
    # Provide an example using MFCC1 as 'true' and a simple model prediction mean to show usage
    y_true_example = df["MFCC1"].values[:100]
    y_pred_example = np.full_like(y_true_example, y_true_example.mean())
    reg_metrics_example = regression_metrics(y_true_example, y_pred_example)
    print("Lab04 A2 (example using MFCC1):", reg_metrics_example)

    # A3-A5: generate 20 training points and plot decision regions for k=3
    train_X, train_y = generate_2d_train(20)
    plt.figure(figsize=(6,6))
    plt.scatter(train_X[:,0], train_X[:,1], c=train_y, cmap='bwr', edgecolor='k')
    plt.title("Lab04 A3 training scatter")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.show()

    # A4 visualize classification over dense grid
    xxg, yyg, Z = classify_grid_with_knn(train_X, train_y, k=3, grid_step=0.1, bounds=(0,10,0,10))
    plt.figure(figsize=(6,6))
    plt.contourf(xxg, yyg, Z, alpha=0.3)
    plt.scatter(train_X[:,0], train_X[:,1], c=train_y, edgecolor='k')
    plt.title("Lab04 A4 decision regions k=3")
    plt.show()

    # A5 repeat for different k values: show 3 panels for k=1,3,9
    ks = [1,3,9]
    plt.figure(figsize=(15,4))
    for i,k in enumerate(ks,1):
        xxg, yyg, Z = classify_grid_with_knn(train_X, train_y, k=k, grid_step=0.15, bounds=(0,10,0,10))
        plt.subplot(1,3,i)
        plt.contourf(xxg, yyg, Z, alpha=0.3)
        plt.scatter(train_X[:,0], train_X[:,1], c=train_y, edgecolor='k')
        plt.title(f"k={k}")
    plt.suptitle("Lab04 A5: effect of k on decision boundary")
    plt.show()

    # A7: hyperparameter tuning on the earlier two-class split
    gs = find_best_k_via_grid(Xtr, ytr, k_values=list(range(1,12,2)))
    print("Lab04 A7: Best k from GridSearch:", gs.best_params_, "best_score:", gs.best_score_)
```
